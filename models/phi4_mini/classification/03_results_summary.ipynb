# Classification results

Read the saved A/C test metrics from the baseline and context-safe experiments. No GPU is required.


```python
from pathlib import Path
import pandas as pd
from IPython.display import display
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'data/forensic_atomic.csv').is_file() and (p / 'pyproject.toml').is_file()), None)
if ROOT is None:
    raise FileNotFoundError('Run this notebook from inside the forensic_atomic repository.')
BASELINE = ROOT / 'results/phi4_mini/classification/baseline/comparison_A_C.csv'
CONTEXT_SAFE = ROOT / 'results/phi4_mini/classification/context_safe/comparison_clean_test_1377.csv'
for path in (BASELINE, CONTEXT_SAFE):
    if not path.is_file():
        raise FileNotFoundError(f'Missing saved result: {path}')
```
